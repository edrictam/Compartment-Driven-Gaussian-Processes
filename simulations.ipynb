# Simulations and illustrative figures

Run this notebook from top to bottom to generate all nine simulation scenarios, the combined performance tables, and the four simulation figures. Each scenario starts its own random stream from seed 2024 plus its scenario offset; the illustrative figures use their own documented seeds.

The production settings are 50 datasets × 10 patients per scenario and 2,000 posterior paths. No saved draws or fitted results are inputs. Aggregation reads back only the summaries written earlier in this run. Run `python reproduce.py --output /tmp/pkgp-fresh` to execute this and all three clinical notebooks in fresh processes.

## Setup and production settings

```python
import os, sys, warnings
from pathlib import Path
warnings.filterwarnings('ignore')
CODE_DIR = Path(os.environ.get('PKGP_CODE_DIR', Path.cwd())).expanduser().resolve()
if not (CODE_DIR / 'scenarios.py').is_file():
    raise RuntimeError('Run from code/ or set PKGP_CODE_DIR to its absolute path.')
if str(CODE_DIR) not in sys.path:
    sys.path.insert(0, str(CODE_DIR))
OUTPUT_DIR = Path(os.environ.get('PKGP_OUTPUT_DIR', CODE_DIR)).expanduser().resolve()
os.environ['PKGP_OUTPUT_DIR'] = str(OUTPUT_DIR)
DATA_DIR = CODE_DIR / 'data'
RES_DIR = OUTPUT_DIR / 'results'
FIG_DIR = OUTPUT_DIR / 'figures'
RES_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)
from scenarios import SCENARIOS, run_scenario, summarize_scenario

settings = {'B': 50, 'N': 10, 'n_paths': 2000, 'seed': 2024, 'kernel': 'matern52', 'sigma_f_scale': 0.6, 'sigma_eps_scale': 0.2, 'ell_fixed': 4.0, 'prior_log_sd': 0.2}
```

## Scenario S1

Correctly specified one-compartment oral, standard 10-sample design.

```python
sc = SCENARIOS['S1']
sc.name, sc.generating_model, sc.fitted_mean, sc.sampling, list(sc.grid)
```

```python
import time
t0 = time.time()
df = run_scenario(sc, **settings, verbose=False)
print(f'elapsed: {time.time() - t0:.1f}s ({len(df)} patient-method rows)')
df.head()
```

```python
summ = summarize_scenario(df)
summ.round(3)
```

```python
df.to_csv(RES_DIR / 'S1_raw.csv', index=False)
summ.to_csv(RES_DIR / 'S1_summary.csv', index=False)
print('saved')
```

## Combined results and simulation figures

Every scenario is required. Read each freshly written summary in S1–S9 order before writing the combined table.

```python
import pandas as pd
import numpy as np
```

## Load all scenario summaries

```python
scenarios = [f'S{i}' for i in range(1, 10)]
missing = [s for s in scenarios if not (RES_DIR / f'{s}_summary.csv').is_file()]
if missing:
    raise FileNotFoundError(f'Missing scenario summaries: {missing}')
summaries = [pd.read_csv(RES_DIR / f'{s}_summary.csv') for s in scenarios]
summary_all = pd.concat(summaries, ignore_index=True)
summary_all.to_csv(RES_DIR / 'summary_all.csv', index=False)
summary_all.shape
```

## AUC performance table

```python
summary_all[['scenario','method','auc_bias','auc_rmse','auc_coverage','auc_width']].round(3)
```

## Cmax performance table

```python
summary_all[['scenario','method','cmax_bias','cmax_rmse','cmax_coverage','cmax_width']].round(3)
```
